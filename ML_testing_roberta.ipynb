# Linear Regression

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GridSearchCV, cross_val_score, KFold
from sklearn.metrics import mean_squared_error, r2_score

# Function to load and preprocess the dataset
def load_and_preprocess_data(file_path, target_column):
    df = pd.read_csv(file_path)
    print("Dataset loaded successfully!")
    print(df.head())
    print(df.columns)
    X = df.drop(columns=[target_column]).values
    y = df[target_column].values
    return X, y, df

# Function to normalize the data
def normalize_data(X):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    return X_scaled

# Function to perform grid search (for extensibility)
def perform_grid_search(model, X, y, cv_folds):
    grid_params = {}  # Linear Regression has no hyperparameters
    grid_search = GridSearchCV(model, grid_params, cv=cv_folds, scoring='neg_mean_squared_error')
    grid_search.fit(X, y)
    return grid_search.best_estimator_

# Function to evaluate the model using cross-validation
def evaluate_model(model, X, y, cv_folds):
    mse_scores = cross_val_score(model, X, y, cv=cv_folds, scoring='neg_mean_squared_error')
    r2_scores = cross_val_score(model, X, y, cv=cv_folds, scoring='r2')
    mse = -mse_scores.mean()  # Convert negative MSE to positive
    rmse = np.sqrt(mse)
    r2 = r2_scores.mean()
    std_dev = np.std(mse_scores)  # Standard deviation of MSE
    return mse, rmse, r2, std_dev

# Main function
def main():
    # Load the data
    file_path = 'roberta_test_2.csv'
    target_column = 'output'
    X, y, df = load_and_preprocess_data(file_path, target_column)

    # Normalize the data
    X_scaled = normalize_data(X)

    # Initialize model and KFold
    model = LinearRegression()
    kf = KFold(n_splits=10, shuffle=True, random_state=42)

    # Perform grid search and get the best model
    best_model = perform_grid_search(model, X_scaled, y, kf)
    print("Best model obtained!")

    # Reduce dataset size for sampling (optional)
    df_sample = df.sample(n=500, random_state=42)
    X_sample = df_sample.drop(columns=[target_column]).values
    y_sample = df_sample[target_column].values

    # Normalize the sample data
    X_sample_scaled = normalize_data(X_sample)

    # Perform grid search and evaluation on the sample
    best_model_sample = perform_grid_search(model, X_sample_scaled, y_sample, kf)
    mse, rmse, r2, std_dev = evaluate_model(best_model_sample, X_sample_scaled, y_sample, kf)

    # Display metrics
    print(f"Metrics for the sample dataset:")
    print(f"MSE: {mse}")
    print(f"RMSE: {rmse}")
    print(f"R²: {r2}")
    print(f"Standard Deviation of MSE: {std_dev}")

if __name__ == "__main__":
    main()
```

# Ridge Regression

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV, cross_val_score, KFold

# Function to load and preprocess the dataset
def load_and_preprocess_data(file_path, target_column):
    df = pd.read_csv(file_path)
    print("Dataset loaded successfully!")
    print(df.head())
    print(df.columns)
    X = df.drop(columns=[target_column]).values
    y = df[target_column].values
    return X, y, df

# Function to normalize the data
def normalize_data(X):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    return X_scaled

# Function to perform grid search for Ridge regression
def perform_grid_search_ridge(X, y, cv_folds, alpha_values):
    model = Ridge()
    grid_params = {'alpha': alpha_values}
    grid_search = GridSearchCV(model, grid_params, cv=cv_folds, scoring='neg_mean_squared_error')
    grid_search.fit(X, y)
    return grid_search.best_estimator_, grid_search.best_params_

# Function to evaluate the model using cross-validation
def evaluate_model(model, X, y, cv_folds):
    mse_scores = cross_val_score(model, X, y, cv=cv_folds, scoring='neg_mean_squared_error')
    r2_scores = cross_val_score(model, X, y, cv=cv_folds, scoring='r2')
    mse = -mse_scores.mean()  # Convert negative MSE to positive
    rmse = np.sqrt(mse)
    r2 = r2_scores.mean()
    std_dev = np.std(mse_scores)  # Standard deviation of MSE
    return mse, rmse, r2, std_dev

# Main function
def main():
    # Load the data
    file_path = 'roberta_test_2.csv'
    target_column = 'output'
    X, y, df = load_and_preprocess_data(file_path, target_column)

    # Normalize the data
    X_scaled = normalize_data(X)

    # Initialize KFold cross-validator
    kf = KFold(n_splits=10, shuffle=True, random_state=42)

    # Perform Grid Search for Ridge regression
    alpha_values = [0.01, 0.1, 1, 10, 100]
    best_model, best_params = perform_grid_search_ridge(X_scaled, y, kf, alpha_values)
    print(f"Best alpha (regularization strength): {best_params['alpha']}")

    # Reduce dataset size for sampling
    df_sample = df.sample(n=500, random_state=42)
    X_sample = df_sample.drop(columns=[target_column]).values
    y_sample = df_sample[target_column].values

    # Normalize the sample data
    X_sample_scaled = normalize_data(X_sample)

    # Perform Grid Search and Cross Validation on the sample
    best_model_sample, _ = perform_grid_search_ridge(X_sample_scaled, y_sample, kf, alpha_values)
    mse, rmse, r2, std_dev = evaluate_model(best_model_sample, X_sample_scaled, y_sample, kf)

    # Display metrics
    print(f"Metrics for the sample dataset:")
    print(f"MSE: {mse}")
    print(f"RMSE: {rmse}")
    print(f"R²: {r2}")
    print(f"Standard Deviation of MSE: {std_dev}")

if __name__ == "__main__":
    main()

```

# KNN

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import GridSearchCV, train_test_split, cross_val_score
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler

# Function to load and preprocess the dataset
def load_and_preprocess_data(file_path, target_column):
    data = pd.read_csv(file_path)
    X = data.drop(columns=[target_column])
    y = data[target_column]
    return X, y

# Function to perform train-test split
def split_data(X, y, test_size=0.2, random_state=42):
    return train_test_split(X, y, test_size=test_size, random_state=random_state)

# Function to scale the features
def scale_features(X_train, X_test):
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    return X_train_scaled, X_test_scaled, scaler

# Function to set up and perform grid search
def perform_grid_search(X_train_scaled, y_train, param_grid):
    knn = KNeighborsRegressor()
    grid_search = GridSearchCV(estimator=knn, param_grid=param_grid, cv=10, scoring='neg_mean_squared_error', verbose=3)
    grid_search.fit(X_train_scaled, y_train)
    return grid_search.best_estimator_, grid_search.best_params_

# Function to evaluate the model on the test set
def evaluate_model(model, X_test_scaled, y_test):
    y_pred_test = model.predict(X_test_scaled)
    mse_test = mean_squared_error(y_test, y_pred_test)
    rmse_test = np.sqrt(mse_test)
    std_dev_test = np.std(y_pred_test)
    r2_test = r2_score(y_test, y_pred_test)
    
    return mse_test, rmse_test, std_dev_test, r2_test

# Function to perform 10-fold cross-validation
def cross_validate_model(model, X, y):
    cv_scores = cross_val_score(model, X, y, cv=10, scoring='neg_mean_squared_error')
    cv_rmse = np.sqrt(-cv_scores)
    cv_mse = -cv_scores
    return cv_mse, cv_rmse

# Main function to run all tasks
def main():
    # Load and preprocess data
    file_path = 'roberta_test_2.csv'
    target_column = 'output'
    X, y = load_and_preprocess_data(file_path, target_column)

    # Split the data into training and testing sets
    X_train, X_test, y_train, y_test = split_data(X, y)

    # Scale the features
    X_train_scaled, X_test_scaled, scaler = scale_features(X_train, X_test)

    # Define parameter grid for KNN
    param_grid = {
        'n_neighbors': [3, 5, 7, 9, 11],
        'weights': ['uniform', 'distance'],
        'algorithm': ['auto', 'ball_tree', 'kd_tree', 'brute'],
        'p': [1, 2]
    }

    # Perform grid search to find the best model
    best_knn, best_params = perform_grid_search(X_train_scaled, y_train, param_grid)

    # Print the best parameters
    print(f'Best Parameters: {best_params}')

    # Evaluate the model on the test set
    mse_test, rmse_test, std_dev_test, r2_test = evaluate_model(best_knn, X_test_scaled, y_test)

    # Print evaluation metrics
    print(f'Mean Squared Error (MSE) on Test Set: {mse_test:.4f}')
    print(f'Root Mean Squared Error (RMSE) on Test Set: {rmse_test:.4f}')
    print(f'Standard Deviation of Predictions on Test Set: {std_dev_test:.4f}')
    print(f"R² value on Test Set: {r2_test:.4f}")

    # Perform 10-fold cross-validation
    cv_mse, cv_rmse = cross_validate_model(best_knn, scaler.fit_transform(X), y)

    # Print cross-validation metrics
    print(f'Cross-Validation MSE: {cv_mse.mean():.4f} ± {cv_mse.std():.4f}')
    print(f'Cross-Validation RMSE: {cv_rmse.mean():.4f} ± {cv_rmse.std():.4f}')

if __name__ == "__main__":
    main()

```

# Polynomial Regression

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import GridSearchCV, train_test_split, KFold
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, r2_score

# Function to load and preprocess the dataset
def load_and_preprocess_data(file_path, target_column):
    data = pd.read_csv(file_path)
    X = data.drop(columns=[target_column])
    y = data[target_column]
    return X, y

# Function to perform train-test split
def split_data(X, y, test_size=0.2, random_state=42):
    return train_test_split(X, y, test_size=test_size, random_state=random_state)

# Function to create the model pipeline
def create_pipeline():
    pipeline = Pipeline([
        ('scaler', StandardScaler()),  # Standardize features
        ('poly', PolynomialFeatures()),  # Polynomial features transformation
        ('linear', LinearRegression())  # Linear regression model
    ])
    return pipeline

# Function to perform grid search for hyperparameter tuning
def perform_grid_search(pipeline, X_train, y_train, param_grid, cv):
    grid_search = GridSearchCV(pipeline, param_grid, cv=cv, scoring='neg_mean_squared_error', verbose=2)
    grid_search.fit(X_train, y_train)
    return grid_search

# Function to evaluate the model on the test set
def evaluate_model(model, X_test, y_test):
    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    std_dev = np.std(y_pred)
    return rmse, r2, std_dev

# Main function to orchestrate the workflow
def main():
    # Load and preprocess the data
    file_path = 'roberta_test_2.csv'
    target_column = 'output'
    X, y = load_and_preprocess_data(file_path, target_column)

    # Split the data into training and test sets
    X_train, X_test, y_train, y_test = split_data(X, y)

    # Create the pipeline
    pipeline = create_pipeline()

    # Define parameter grid for GridSearchCV
    param_grid = {
        'poly__degree': [2, 3, 4],  # Different degrees of polynomial features
        'linear__fit_intercept': [True, False]  # Fit intercept or not
    }

    # Set up cross-validation
    cv = KFold(n_splits=10, shuffle=True, random_state=42)

    # Perform grid search to find the best model
    grid_search = perform_grid_search(pipeline, X_train, y_train, param_grid, cv)

    # Get the best model
    best_model = grid_search.best_estimator_

    # Print the best parameters
    print(f'Best Parameters: {grid_search.best_params_}')

    # Evaluate the model on the test set
    rmse, r2, std_dev = evaluate_model(best_model, X_test, y_test)

    # Print evaluation metrics
    print(f'Root Mean Squared Error (RMSE) on Test Set: {rmse}')
    print(f'R² value on Test Set: {r2}')
    print(f'Standard Deviation of Predictions: {std_dev}')

if __name__ == "__main__":
    main()
```

# Logistic Regression

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler

# Function to load and preprocess the dataset
def load_and_preprocess_data(file_path, target_column):
    df = pd.read_csv(file_path)
    X = df.drop(columns=[target_column])
    y = df[target_column]
    return X, y

# Function to standardize the features
def standardize_features(X):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    return X_scaled

# Function to split data into training and testing sets
def split_data(X, y, test_size=0.2, random_state=42):
    return train_test_split(X, y, test_size=test_size, random_state=random_state)

# Function to create the Logistic Regression model
def create_logistic_regression_model():
    return LogisticRegression(max_iter=1000)

# Function to perform grid search for hyperparameter tuning
def perform_grid_search(model, X_train, y_train, param_grid, cv=10):
    grid_search = GridSearchCV(estimator=model, param_grid=param_grid, cv=cv, scoring='neg_mean_squared_error')
    grid_search.fit(X_train, y_train)
    return grid_search

# Function to evaluate the model on the test set
def evaluate_model(model, X_test, y_test):
    y_pred = model.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    std_dev = np.std(y_pred)
    return r2, rmse, std_dev

# Function for cross-validation and calculating RMSE
def cross_val_rmse(model, X_train, y_train, cv=10):
    cross_val_scores = cross_val_score(model, X_train, y_train, cv=cv, scoring='neg_mean_squared_error')
    cv_rmse = np.mean(np.sqrt(-cross_val_scores))
    return cv_rmse

# Main function to orchestrate the workflow
def main():
    # Load and preprocess the data
    file_path = 'roberta_test_2.csv'
    target_column = 'output'
    X, y = load_and_preprocess_data(file_path, target_column)

    # Standardize the features
    X_scaled = standardize_features(X)

    # Split the data into training and testing sets
    X_train, X_test, y_train, y_test = split_data(X_scaled, y)

    # Create the Logistic Regression model
    model = create_logistic_regression_model()

    # Define the parameter grid for GridSearchCV
    param_grid = {
        'C': [0.01, 0.1, 1, 10, 100],
        'solver': ['newton-cg', 'lbfgs', 'liblinear'],
        'penalty': ['l2'],  # l2 regularization for Logistic Regression
    }

    # Perform grid search to find the best parameters
    grid_search = perform_grid_search(model, X_train, y_train, param_grid)

    # Print the best parameters
    print("Best parameters found: ", grid_search.best_params_)

    # Get the best model from grid search
    best_model = grid_search.best_estimator_

    # Evaluate the model on the test set
    r2, rmse, std_dev = evaluate_model(best_model, X_test, y_test)

    # Print evaluation metrics
    print("R² Score: ", r2)
    print("RMSE: ", rmse)
    print(f'Standard Deviation of Predictions: {std_dev}')

    # Cross-validation RMSE
    cv_rmse = cross_val_rmse(best_model, X_train, y_train)
    print("Cross-validated RMSE: ", cv_rmse)

if __name__ == "__main__":
    main()
```

# Decision Tree Regression

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, RandomizedSearchCV, cross_val_score, KFold
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.tree import DecisionTreeRegressor

# Function to load and preprocess the dataset
def load_and_preprocess_data(file_path, target_column):
    df = pd.read_csv(file_path)
    X = df.drop(columns=[target_column])
    y = df[target_column]
    return X, y

# Function to standardize the features
def standardize_features(X):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    return X_scaled

# Function to apply PCA for dimensionality reduction
def apply_pca(X_scaled, variance_threshold=0.95):
    pca = PCA(n_components=variance_threshold)
    X_pca = pca.fit_transform(X_scaled)
    return X_pca

# Function to split the data into training and testing sets
def split_data(X, y, test_size=0.2, random_state=42):
    return train_test_split(X, y, test_size=test_size, random_state=random_state)

# Function to define the model
def create_decision_tree_model():
    return DecisionTreeRegressor()

# Function for RandomizedSearchCV for hyperparameter tuning
def perform_randomized_search(model, X_train, y_train, param_dist, n_iter=10, cv=5, random_state=42):
    random_search = RandomizedSearchCV(model, param_distributions=param_dist, n_iter=n_iter, cv=cv, random_state=random_state)
    random_search.fit(X_train, y_train)
    return random_search

# Function for cross-validation
def cross_validate(model, X_train, y_train, cv_splits=5):
    kf = KFold(n_splits=cv_splits, shuffle=True, random_state=42)
    cv_scores = cross_val_score(model, X_train, y_train, cv=kf)
    return cv_scores

# Function to evaluate the model
def evaluate_model(model, X_test, y_test):
    y_pred = model.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)
    mean_value = np.mean(y_test)
    std_dev = np.std(y_test)
    return r2, rmse, mae, mean_value, std_dev

# Function to print the results
def print_results(random_search, r2, rmse, mae, mean_value, std_dev, cv_scores):
    print(f"Best Parameters: {random_search.best_params_}")
    print(f"R^2 Score: {r2}")
    print(f"RMSE: {rmse}")
    print(f"MAE: {mae}")
    print(f"Mean of Output: {mean_value}")
    print(f"Standard Deviation of Output: {std_dev}")
    print(f"Cross-Validation Scores: {cv_scores}")
    print(f"Cross-Validation Mean: {cv_scores.mean()}")
    print(f"Cross-Validation Std Dev: {cv_scores.std()}")

# Main function to orchestrate the workflow
def main():
    # Load and preprocess the data
    file_path = 'roberta_test_2.csv'
    target_column = 'output'
    X, y = load_and_preprocess_data(file_path, target_column)

    # Standardize the features
    X_scaled = standardize_features(X)

    # Apply PCA for dimensionality reduction
    X_pca = apply_pca(X_scaled)

    # Split the data into training and testing sets
    X_train, X_test, y_train, y_test = split_data(X_pca, y)

    # Define the Decision Tree model
    model = create_decision_tree_model()

    # Define the parameter grid for RandomizedSearchCV
    param_dist = {
        'max_depth': [None, 10, 20, 30, 40, 50],
        'min_samples_split': np.arange(2, 10),
        'min_samples_leaf': np.arange(1, 5),
        'max_features': ['auto', 'sqrt', 'log2', None]
    }

    # Perform RandomizedSearchCV for hyperparameter tuning
    random_search = perform_randomized_search(model, X_train, y_train, param_dist)

    # Perform cross-validation
    cv_scores = cross_validate(random_search.best_estimator_, X_train, y_train)

    # Evaluate the model on the test set
    r2, rmse, mae, mean_value, std_dev = evaluate_model(random_search.best_estimator_, X_test, y_test)

    # Print the results
    print_results(random_search, r2, rmse, mae, mean_value, std_dev, cv_scores)

if __name__ == "__main__":
    main()
```
